```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('iphone.csv')
#The column names have spaces . rename the column names to have underscore '_' instead of space
df.columns=df.columns.str.replace(' ','_')
print(df.columns)

print(df.describe())
```

```python
#start rating for some of the models is missing in the dataset.
#fill those missing values with the average rating all the models.


# Calculate the average rating excluding None (missing values)
average_rating=df['Star_Rating'].mean(skipna=True)
print(average_rating)

# Fill missing values with the calculated average rating
df['Star_Rating'].fillna(average_rating, inplace=True)
print(df)
```

```python
#Now instead of filling missing values with avg rating of full dataset , fill with avg rating based on RAM
# Calculate the average rating for each RAM group
average_rating_by_RAM=df.groupby('Ram')['Star_Rating'].mean().reset_index()
print(average_rating_by_RAM)

# Rename the 'Star Rating' column to 'Average Rating' for clarity
average_rating_by_RAM.rename(columns={'Star_Rating':'Average_Rating'},inplace=True)
print(average_rating_by_RAM.columns)


```

```python
#create a new column in the dataframe "Discount_Percentage" based on MRP and sale value
df['Discount_Percentage']=((df['Mrp']-df['Sale_Price'])/df['Mrp'])*100
print(df['Discount_Percentage'])
```

```python
#which model has highest percent discount ?
sorted_df=df.sort_values(by='Discount_Percentage',ascending=False)
print(sorted_df)

# Get the model with the highest discount and its corresponding percentage
model_with_highest_discount=sorted_df.iloc[0]['Product_Name']
print(model_with_highest_discount)

highest_disount_percentage=sorted_df.iloc[0]['Discount_Percentage']
print(highest_disount_percentage)
```

```python
#find total no of models  each space configuration (128 GB , 64 GB etc)
space_configuration_counts=df['Ram'].value_counts()
print(space_configuration_counts)

# Rename the columns for clarity
space_configuration_counts = space_configuration_counts.reset_index()
space_configuration_counts.rename(columns={'Space_Configuration':'Total_Models'})
```

```python
 #find total number of models for each color

# Define a function to extract color based on a pattern
def extract_color(Product_Name):
  parts=Product_Name.split('(')
  if len(parts)>1:
    color_part=parts[1].strip()
    if ',' in color_part:
      color=color_part.split(',')[0].strip()
      return color
  return None

# Apply the extract_color function to create a 'Color' column
df['color']=df['Product_Name'].apply(extract_color)
print(df['color'])

#Count the number of models for each color
color_counts=df['color'].value_counts()
print(color_counts)
```

```python
 #find total number of models by iphone version :
 # Define a function to extract iPhone version
def extract_iphone_version(Product_Name):
  parts=Product_Name.lower().split('iphone')
  if len(parts)>1:
    version=parts[1].strip().split(' ')[0]
    return 'iphone'+version
  return None

df['iphone_version'] = df['Product_Name'].apply(extract_iphone_version)

print(df['iphone_version'])

# Count the number of models for each iPhone version
iphone_version_counts = df['iphone_version'].value_counts().reset_index()

print(iphone_version_counts)


```

```python
#list top 5 models having highest no of reviews

# Sort the DataFrame by 'Number Of Reviews' in descending order
df_sorted=df.sort_values(by='Number_Of_Reviews',ascending=False)
print(df_sorted)

top_5_models=df_sorted.head(5)
print(top_5_models)
```

```python
#what is the price diffrence between highest price and lowest price iphone (based on mrp)

# Find the highest and lowest MRP values
highest_price=df['Mrp'].max()
lowest_price=df['Mrp'].min()
price_difference=highest_price-lowest_price
print(price_difference)
```

```python
#which iphone has 3rd highest MRP

# Sort the DataFrame by 'Mrp' column in descending order and reset the index
sorted_df=df.sort_values(by='Mrp',ascending=False).reset_index(drop=True)
# Get the iPhone with the 3rd highest MRP
third_highest_mrp=sorted_df.loc[2]
print(third_highest_mrp)
```

```python
 #what is the average mrp of iphones which costs above 100,000

# Filter the DataFrame to include only iPhones with MRP above 100,000
high_mrp_iphone=df[df['Mrp']>100000]
print(high_mrp_iphone)

# Calculate the average MRP for the filtered iPhones
average_mrp=high_mrp_iphone['Mrp'].mean()
print(average_mrp)
```

```python
#which iphone with 128 GB space has highest ratings to review ratio

# Filter the DataFrame to include only iPhones with 128 GB of space
iphone_128gb=df[df['Ram']=='128 GB']
print(iphone_128gb.shape)


iphone_128gb = df[
    df['Product_Name'].str.contains('iphone', case=False, na=False) &
    df['Product_Name'].str.contains('128 GB', case=False, na=False)
]

print(iphone_128gb[['Product_Name']])

iphone_128gb['Ratings_to_review'] = (
    iphone_128gb['Number_Of_Ratings'] /
    iphone_128gb['Number_Of_Reviews']
)

highest_ratio_iphone = iphone_128gb.loc[
    iphone_128gb['Ratings_to_review'].idxmax()
]

print(highest_ratio_iphone)
```

```python
#find total no of reviews for iphone 11 and iphone 12 category .
#Output should have only 2 rows (for 11 and 12).

# Filter the DataFrame for iPhone 11 and iPhone 12 categories
iphone_11_reviews = df[df['Product_Name'].str.contains('iPhone 11', case=False)]
iphone_12_reviews = df[df['Product_Name'].str.contains('iPhone 12', case=False)]

# Calculate the total number of reviews for each category
total_reviews_11 = iphone_11_reviews['Number_Of_Reviews'].sum()
total_reviews_12 = iphone_12_reviews['Number_Of_Reviews'].sum()

# Create a new DataFrame with the results
result_df = pd.DataFrame({
    'iPhone Category': ['iPhone 11', 'iPhone 12'],
    'Total Reviews': [total_reviews_11, total_reviews_12]
})
print(result_df)

```
